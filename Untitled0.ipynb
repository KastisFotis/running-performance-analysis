```python
def convert_distance(row):
  if row['Τύπος δραστηριότητας'] == 'Τρέξιμο στίβου':
    meters = float(row['Απόσταση'].replace(',', ''))
    return meters / 1000
  else:
    return float(row['Απόσταση'])

df['distance_km'] = df.apply(convert_distance, axis=1)
```
